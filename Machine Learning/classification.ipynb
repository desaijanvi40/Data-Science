```python
# ========================================== 
# INTRODUCTION TO CLASSIFICATION 
# REAL-LIFE EXAMPLE: Student Pass/Fail Prediction 
# ========================================== 
# Step 1: Import libraries 
import pandas as pd 
# Step 2: Create dataset 
data = { 
"Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8], 
"Result": ["Fail", "Fail", "Fail", "Pass", "Pass", "Pass", "Pass", "Pass"] 
} 
df = pd.DataFrame(data) 
print("===== DATASET =====") 
print(df) 
# Step 3: Identify input and output 
X = df[["Study_Hours"]] 
y = df["Result"] 
print("\n===== INPUT (X) =====") 
print(X) 
print("\n===== OUTPUT (y) =====") 
print(y) 
# Step 4: Check unique classes 
print("\n===== UNIQUE CLASSES =====") 
print(y.unique()) 
# Step 5: Count class distribution 
print("\n===== CLASS COUNT =====") 
print(y.value_counts()) 
# Step 6: Identify type of classification 
num_classes = len(y.unique()) 
if num_classes == 2: 
    print("\nThis is a BINARY CLASSIFICATION problem") 
else: 
print("\nThis is a MULTICLASS CLASSIFICATION problem")
```
